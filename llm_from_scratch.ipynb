Text - numbers - mathematical operations - probabilities - prediction. P(next token | previous tokens)

The probability of the next token given the tokens we've already seen . the model(a mathematical function/computation) computes a probability distribution over its entire vocabulary, and they all add up to 1.

The loss function is how the model knows it's prediction is wrong. suppose the previous tokens are 'the dog ' and the answer is 'sat' but the model predicts:
sat - 0.05
ate - 0.80
dog - 0.10
....
L = -log(p)
if p = 1, then L = -log(1) = 0. if p = 0.1, L = -log(0.1) -log(10^-1) = 1

The gradient determines the direction of the update.
The learning rate determines how large the step is in that direction.

1. Forward pass
Calculate the prediction and loss.
2. Backpropagation
Use the chain rule to calculate gradients.
3. Gradient descent


```python
# Our training dataset

x = 2.0
y = 10.0
w = 3.0

# Make prediction: forward pass
prediction = w * x

# Calculate the loss
loss = (prediction - y) ** 2

print("Prediction: ", prediction)
print("Loss: ", loss)
```

```python
# Calculate the gradient

gradient = 2 * (prediction - y) * x

# Choose the learning rate
learning_rate = 0.1

# Update the weight
new_w = w - learning_rate * gradient

print("Gradient:", gradient)
print("Old Weight:", w)
print("New Weight:", new_w)
```

```python
# New prediction
new_prediction = new_w * x

# Calculate loss
new_loss = (new_prediction - y)** 2
print("New prediction:", new_prediction)
print("New loss:", new_loss)
```

```python
# iterative optimization.
# Reset the model parameter
w = 3

# Training settings
learning_rate = 0.1
epochs = 10

for epoch in range(epochs):
  # Forward pass
  prediction = w * x

  # Calculate loss
  loss = (prediction - y) ** 2

  # Calculate gradient
  gradient = 2 * (prediction - y) * x
  # Update weight
  w = w - learning_rate * gradient
```
