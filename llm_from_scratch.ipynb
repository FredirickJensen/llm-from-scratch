Text - numbers - mathematical operations - probabilities - prediction. P(next token | previous tokens)

The probability of the next token given the tokens we've already seen . the model(a mathematical function/computation) computes a probability distribution over its entire vocabulary, and they all add up to 1.

The loss function is how the model knows it's prediction is wrong. suppose the previous tokens are 'the dog ' and the answer is 'sat' but the model predicts:
sat - 0.05
ate - 0.80
dog - 0.10
....
L = -log(p)
if p = 1, then L = -log(1) = 0. if p = 0.1, L = -log(0.1) -log(10^-1) = 1

The gradient determines the direction of the update.
The learning rate determines how large the step is in that direction.

1. Forward pass
Calculate the prediction and loss.
2. Backpropagation
Use the chain rule to calculate gradients.
3. Gradient descent


```python
# Our training dataset

x = 2.0
y = 10.0
w = 3.0

# Make prediction: forward pass
prediction = w * x

# Calculate the loss
loss = (prediction - y) ** 2

print("Prediction: ", prediction)
print("Loss: ", loss)
```

```python
# Calculate the gradient

gradient = 2 * (prediction - y) * x

# Choose the learning rate
learning_rate = 0.1

# Update the weight
new_w = w - learning_rate * gradient

print("Gradient:", gradient)
print("Old Weight:", w)
print("New Weight:", new_w)
```

```python
# New prediction
new_prediction = new_w * x

# Calculate loss
new_loss = (new_prediction - y)** 2
print("New prediction:", new_prediction)
print("New loss:", new_loss)
```

```python
# iterative optimization.
# Reset the model parameter
w = 3

# Training settings
learning_rate = 0.1
epochs = 10

for epoch in range(epochs):
  # Forward pass
  prediction = w * x

  # Calculate loss
  loss = (prediction - y) ** 2

  # Calculate gradient
  gradient = 2 * (prediction - y) * x
  # Update weight
  w = w - learning_rate * gradient
```

```python

final_prediction = w * x
final_loss = (final_prediction - y) ** 2

print("Final weight:", w)
print("Final prediction:", final_prediction)
print("Target:", y)
print("Final loss:", final_loss)
```

```python
# Training exammple
x = 2.0
y = 10.0

# Initial parameters
w = 3.0
b = 0.0

# Learning rate
learning_rate = 0.1

# Forward pass
prediction = w * x + b
loss = (prediction - y) ** 2

# Calculate both gradients
gradient_w = 2 * (prediction - y) * x
gradient_b = 2 * (prediction - y)

# Update both parameters
w = w - learning_rate * gradient_w
b = b - learning_rate * gradient_b

# Evaluate the updated model
new_prediction = w * x + b
new_loss = (new_prediction - y) ** 2

print("Updated weight:", w)
print("Updated bias:", b)
print("New prediction:", new_prediction)
print("New loss:", new_loss)
```

```python
# Our training dataset
xs = [1.0, 2.0, 3.0]
ys = [5.0, 10.0, 15.0]

# Initial parameters
w = 3.0
b = 0.0

# Training settings
learning_rate = 0.01
epochs = 1000

for epoch in range(epochs):
  total_loss = 0.0
  gradient_w = 0.0
  gradient_b = 0.0

  # Calculate loss and gradients across the examples
  for x_i, y_i in zip(xs, ys):
    prediction = w * x_i + b
    error = prediction - y_i

    total_loss += error ** 2
    gradient_w += 2 * error * x_i
    gradient_b += 2 * error


  # Convert sums into averages
  n = len(xs)
  mse = total_loss / n
  gradient_w /= n
  gradient_b /= n

  # Update parameters
  w -= learning_rate * gradient_w
  b -= learning_rate * gradient_b

  if epoch % 10 == 0:
    print(
        f"Epoch {epoch:3d} | "
        f"MSE: {mse:.4f} | "
        f"w: {w:.4f} | "
        f"b: {b:.4f}"
    )

print("\nFinal parameters:")
print("Weight:", w)
print("Bias:", b)
```

```python
import torch

# Trainable parameters
w = torch.tensor(3.0, requires_grad=True)
b = torch.tensor(1.0, requires_grad=True)
x = torch.tensor(2.0)
y = torch.tensor(10.0)

prediction = w * x + b
loss = (prediction - y) ** 2

loss.backward()

print("Loss", loss.item())
print("Weight gradient:", w.grad.item())
print("Bias gradient:", b.grad.item())
```

```python
import torch

# Training data
xs = torch.tensor([1.0, 2.0, 3.0])
ys = torch.tensor([5.0, 10.0, 15.0])

# Trainable parameters
w = torch.tensor(3.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)

# Optimizer
learning_rate = 0.01
optimizer = torch.optim.SGD([w, b], lr=learning_rate)
```

```python
epochs = 100

for epoch in range(epochs):
  # 1. Clear previous gradients
  optimizer.zero_grad()

  # 2. Forward pass
  prediction = w * xs + b

  # 3. Calculate loss
  loss = ((prediction - ys) ** 2).mean()

  # 4. Backward pass
  loss.backward()

  # 5. Update parameters
  optimizer.step()
```
